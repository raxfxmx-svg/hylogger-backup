```python
from pytsg import parse_tsg
import matplotlib.pyplot as plt

folder = r"C:\path\to\project\8a294cfc-68df-4f03-995b-c5fd0d47ab0"
data = parse_tsg.read_package(folder)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(data.nir.wavelength, data.nir.spectra[0])
plt.title("NIR")

plt.subplot(1, 2, 2)
plt.plot(data.tir.wavelength, data.tir.spectra[0])
plt.title("TIR")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Core depth
depth = data.nir.sampleheaders["D"].astype(float).to_numpy()

# Plot every tenth point to keep rendering responsive.
step = 10
z = data.nir.spectra[::step]
d = depth[::step]

# Exclude extreme values to improve colour contrast.
vmin, vmax = np.nanpercentile(z, [2, 98])

plt.figure(figsize=(14, 8))

plt.pcolormesh(
    data.nir.wavelength,
    d,
    z,
    shading="auto",
    cmap="turbo",
    vmin=vmin,
    vmax=vmax
)

plt.gca().invert_yaxis()
plt.colorbar(label="Reflectance")
plt.xlabel("Wavelength (nm)")
plt.ylabel("Depth (m)")
plt.title("05GJD001 NIR Spectral Heatmap")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

depth = data.nir.sampleheaders["D"].astype(float).to_numpy()

# Target depth
target_depths = [150, 180, 210, 240, 270, 300]

plt.figure(figsize=(12, 6))

for target in target_depths:
    # Find the sample closest to the target depth.
    i = np.argmin(np.abs(depth - target))

    plt.plot(
        data.nir.wavelength,
        data.nir.spectra[i],
        label=f"{depth[i]:.2f} m"
    )

plt.xlabel("Wavelength (nm)")
plt.ylabel("Reflectance")
plt.title("NIR Spectra at Different Depths")
plt.legend(title="Depth")
plt.grid(alpha=0.2)
plt.show()
```
